IMPORTING THE LIBRARIES

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('The Rocket propellant Data.csv')
df
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
x=np.array(df['Shear strength'])
y=np.array(df['Age of propellant'])
```

```python
plt.scatter(x,y)
plt.xlabel('Shear strength')
plt.ylabel('Age of propellant')
plt.title('Scatter Plot')
plt.show()
```

```python
x.mean()
```

```python
y.mean()
```

```python
plt.scatter(x.mean(),y.mean(),color='red')
plt.scatter(x,y)
plt.xlabel('Shear strength')
plt.ylabel('Age of propellant')
plt.title('Scatter Plot')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg=LinearRegression()
```

```python
reg.fit(x.reshape(-1,1),y)
```

```python
m=reg.coef_
```

```python
c=reg.intercept_
```

```python
m*2053+c
```

```python
reg.predict([[2053]])
```

```python
df['predicted_y']=reg.predict(x.reshape(-1,1))
df
```

```python
df['loss']=df['Age of propellant']-df['predicted_y']
df['loss']
```

```python
plt.plot(x.reshape(-1, 1), reg.predict(x.reshape(-1, 1)),color='red')
plt.scatter(x.mean(), y.mean(), color='red', marker='o')

plt.scatter(x,y, color='green',marker='.')

plt.xlabel('Shear strength')
plt.ylabel('Age of propellant')
plt.title('Scatter Plot')
plt.show()
```

```python
plt.scatter(x.reshape(-1,1), reg.predict(x.reshape(-1,1)), color='black',marker='.',s=100)
plt.plot(x.reshape(-1, 1), reg.predict(x.reshape(-1, 1)),color='red')
plt.scatter(x.mean(), y.mean(), color='red', marker='o')

plt.scatter(x,y, color='green',marker='.')

plt.xlabel('Shear strength')
plt.ylabel('Age of propellant')
plt.title('Scatter Plot')
plt.show()
```

```python
from sklearn.metrics import r2_score
```

```python
r2=r2_score(df['Age of propellant'],df['predicted_y'])
r2
```
