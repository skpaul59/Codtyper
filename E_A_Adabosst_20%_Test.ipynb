# Task
Analyze a materials science dataset focusing on material toughness. The analysis involves several steps: loading and inspecting the data, visualizing correlations, training and evaluating an XGBoost regression model, visualizing feature importances, plotting actual vs. predicted values, generating a residual plot, and finally saving the trained model. The ultimate goal is to understand the factors influencing material toughness and to build a predictive model for it.

## Load Energy Absorption Data

### Subtask:
Load the 'property_Energy_Absorption.csv' dataset into a pandas DataFrame.


**Reasoning**:
I need to load the 'property_Energy_Absorption.csv' file into a pandas DataFrame and display its head as per the instructions.



```python
import pandas as pd

# Load the dataset
df_energy = pd.read_csv('/content/property_Energy_Absorption.csv')

# Display the first 5 rows of the DataFrame
display(df_energy.head())
```

## Initial Data Inspection (Energy Absorption)

### Subtask:
Display DataFrame information (`df.info()`) and descriptive statistics (`df.describe()`) for the 'Energy Absorption' dataset to understand its structure and content.


**Reasoning**:
As per the subtask instructions, I need to display the DataFrame information and descriptive statistics for `df_energy` to understand its structure and content.



```python
print('--- DataFrame Info ---')
df_energy.info()

print('\n--- Descriptive Statistics ---')
display(df_energy.describe())
```

## Correlation Matrix Heatmap (Energy Absorption)

### Subtask:
Generate a heatmap to visualize the correlations between numerical features and the 'Energy Absorption' target in the dataset.


**Reasoning**:
I need to calculate the correlation matrix for numerical columns in `df_energy`, create a heatmap, add a title, and then display the plot.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix for numerical columns including 'Energy_Absorption'
correlation_matrix_energy = df_energy.select_dtypes(include=['number']).corr()

# Create a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix_energy, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Numerical Features (Energy Absorption)')
plt.show()
```

## Prepare Data for Modeling (Energy Absorption)

### Subtask:
Separate features (X) and the new target (y = 'Energy Absorption'), drop the 'Sample_ID' column, check for missing values, and split the data into training and testing sets.


**Reasoning**:
I need to prepare the data for modeling by separating features and target, dropping the 'Sample_ID' column, checking for missing values, and splitting the data into training and testing sets.



```python
from sklearn.model_selection import train_test_split

# Separate features (X) and target (y)
X_energy = df_energy.drop(['Energy_Absorption', 'Sample_ID'], axis=1)
y_energy = df_energy['Energy_Absorption']

# Check for missing values
print('''Missing values in features (Energy Absorption):
''', X_energy.isnull().sum())
print('''
Missing values in target (Energy Absorption):
''', y_energy.isnull().sum())

# Split the data into training and testing sets
X_train_energy, X_test_energy, y_train_energy, y_test_energy = train_test_split(X_energy, y_energy, test_size=0.2, random_state=42)

print('''
X_train_energy shape:''', X_train_energy.shape)
print('X_test_energy shape:', X_test_energy.shape)
print('y_train_energy shape:', y_train_energy.shape)
print('y_test_energy shape:', y_test_energy.shape)
```

## Train and Evaluate XGBoost Model (Energy Absorption)

### Subtask:
Initialize and train an XGBoost Regressor model for 'Energy Absorption', perform hyperparameter tuning using GridSearchCV, and evaluate its performance using MAE, MSE, and R2 score.


**Reasoning**:
I need to initialize and train an XGBoost Regressor model for 'Energy Absorption', perform hyperparameter tuning using GridSearchCV, and evaluate its performance using MAE, MSE, and R2 score, as outlined in the subtask instructions. This code block will perform all these steps.



```python
from xgboost import XGBRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Initialize the XGBoost Regressor
xgb_energy = XGBRegressor(random_state=42)

# Define the parameter grid for hyperparameter tuning for XGBoost for Energy Absorption
param_grid_energy = {
    'n_estimators': [50, 100, 150],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 4, 5],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0]
}

# Perform GridSearchCV for hyperparameter tuning
grid_search_xgb_energy = GridSearchCV(estimator=xgb_energy, param_grid=param_grid_energy,
                                  cv=5, n_jobs=-1, verbose=2, scoring='neg_mean_squared_error')
grid_search_xgb_energy.fit(X_train_energy, y_train_energy)

# Get the best estimator
best_xgb_energy = grid_search_xgb_energy.best_estimator_

print(f"Best Hyperparameters for XGBoost (Energy Absorption): {grid_search_xgb_energy.best_params_}")

# Make predictions on the test set
y_pred_xgb_energy = best_xgb_energy.predict(X_test_energy)

# Evaluate the model
mae_xgb_energy = mean_absolute_error(y_test_energy, y_pred_xgb_energy)
mse_xgb_energy = mean_squared_error(y_test_energy, y_pred_xgb_energy)
r2_xgb_energy = r2_score(y_test_energy, y_pred_xgb_energy)

print(f"\nXGBoost Mean Absolute Error (MAE) for Energy Absorption: {mae_xgb_energy:.4f}")
print(f"XGBoost Mean Squared Error (MSE) for Energy Absorption: {mse_xgb_energy:.4f}")
print(f"XGBoost R-squared (R2) Score for Energy Absorption: {r2_xgb_energy:.4f}")
```

## Visualize Feature Importances (Energy Absorption)

### Subtask:
Generate a bar chart showing the importance of each feature in the trained XGBoost model for 'Energy Absorption'.


**Reasoning**:
I need to extract feature importances from the trained XGBoost model, create a DataFrame with feature names and importances, sort it, and then plot a horizontal bar chart to visualize these importances, as per the subtask instructions.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get feature importances from the best XGBoost model for Energy Absorption
feature_importances_xgb_energy = best_xgb_energy.feature_importances_

# Get feature names from the training data
feature_names_energy = X_train_energy.columns

# Create a DataFrame for better visualization
importance_df_xgb_energy = pd.DataFrame({'Feature': feature_names_energy, 'Importance': feature_importances_xgb_energy})
importance_df_xgb_energy = importance_df_xgb_energy.sort_values(by='Importance', ascending=False)

# Plot feature importances
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', hue='Feature', data=importance_df_xgb_energy, palette='viridis', legend=False)
plt.title('Feature Importances from Best XGBoost Model for Energy Absorption')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
```

## Plot Actual vs. Predicted Values (Energy Absorption)

### Subtask:
Create a scatter plot to compare the actual 'Energy Absorption' values against the values predicted by the XGBoost model.


**Reasoning**:
I need to create a scatter plot comparing actual and predicted 'Energy Absorption' values, add a diagonal line for perfect prediction, and set appropriate labels and title, as per the subtask instructions.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot of actual vs. predicted values for XGBoost (Energy Absorption)
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test_energy, y=y_pred_xgb_energy) # Using y_pred_xgb_energy for XGBoost predictions

# Add a diagonal line for perfect prediction
plt.plot([y_test_energy.min(), y_test_energy.max()], [y_test_energy.min(), y_test_energy.max()], 'r--')

plt.title('Actual vs. Predicted Energy Absorption Values (XGBoost)')
plt.xlabel('Actual Energy Absorption')
plt.ylabel('Predicted Energy Absorption')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## Generate Residual Plot (Energy Absorption)

### Subtask:
Create a residual plot to analyze the distribution of errors between actual and predicted 'Energy Absorption' values for the XGBoost model.


**Reasoning**:
I need to calculate the residuals and then create a scatter plot to visualize the distribution of these errors against the predicted values, as per the instructions for generating a residual plot.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate residuals
residuals_energy = y_test_energy - y_pred_xgb_energy

# Create a residual plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_pred_xgb_energy, y=residuals_energy)
plt.axhline(y=0, color='r', linestyle='--') # Add a horizontal line at y=0

plt.title('Residual Plot (Predicted vs. Residuals - XGBoost for Energy Absorption)')
plt.xlabel('Predicted Energy Absorption')
plt.ylabel('Residuals (Actual - Predicted)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## Save Trained XGBoost Model (Energy Absorption)

### Subtask:
Save the best-performing XGBoost model trained for 'Energy Absorption' to a file for future use.


**Reasoning**:
I need to save the `best_xgb_energy` model using `joblib.dump()` to a specified filename and then print a confirmation message as per the subtask instructions.



```python
import joblib

# Define the filename for the saved model
model_filename_energy = 'xgboost_energy_absorption_model.joblib'

# Save the best XGBoost model to the file
joblib.dump(best_xgb_energy, model_filename_energy)

print(f"Model saved successfully to {model_filename_energy}")
```
